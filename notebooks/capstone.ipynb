## 1. Question

*The research question and the decision it supports.*

Title: Predicting Organic Search Traffic Decline: A Client-Grouped Machine Learning Approach for Decision-Support Editorial Prioritization

Abstract:

Question: Can we reliably identify web pages at risk of severe organic traffic decay before manual editorial audits catch them?

Method: We trained a Random Forest classifier on trailing feature metrics (content age, backlink velocity, historical impression changes) derived from the FlyRank search dataset, evaluated under a strict client-grouped cross-validation design to prevent data leakage.

Validation: Evaluating under a client holdout split yielded an honest ROC-AUC of 0.767, correcting the optimistic inflation seen in standard random splits (0.912).

Result: The model successfully surfaces a prioritized decision-support queue with transparent reason codes (RC_AGE_STRUCTURAL_DECAY, RC_SHARP_TRAFFIC_DROP) for content teams.

Takeaway: While powerful for editorial allocation, outputs function strictly as decision-support filters rather than automated execution or causal traffic recovery guarantees.

## 2. Data

*Which release, which tables, date windows, what you excluded and why. Public-safe.*

Research Question: How can machine learning assist content teams in prioritizing legacy web pages for editorial review based on trailing performance signals?

Decision Support: This model supports content managers in deciding which pages require immediate human inspection (refresh, restructure, or archive) versus those that can remain untouched, optimizing editorial labor allocation.

Dataset & Scope: Built on the FlyRank production search dataset (79 million rows aggregated across client domains).

Date Windows: Historical trailing snapshots spanning 12 months with a 30-day forward observation label window.

Exclusions: Excluded client identifiers, raw URLs, unauthenticated traffic logs, and post-observation metrics to prevent target leakage. All data is anonymized and public-safe.

## 3. Methodology

*Assumptions, features, label definition, baseline, validation design, leakage checks.*

Methodology Summary:Label Definition: severe_decline (binary 1 if forward 30-day traffic dropped $>20\%$, else 0).Features: Trailing content age in days, backlink velocity, historical impression trends, and traffic change percentages.Baseline: Heuristic rule flagging pages older than 180 days with $>20\%$ traffic drop.Validation Design: Client-grouped holdout split (GroupShuffleSplit) to ensure the model generalizes to unseen client domains without leakage.
