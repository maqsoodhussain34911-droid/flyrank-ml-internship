## 3. Verify it with queries (grain, counts, missing values, windows)

*Every claim above gets a query cell here. A contract claim without a query next to it is a guess.*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
# Verification query for null values and record counts across features and label
query_data_quality = """
SELECT
    COUNT(*) AS total_records,
    SUM(CASE WHEN query_length IS NULL THEN 1 ELSE 0 END) AS null_query_length,
    SUM(CASE WHEN historical_ctr IS NULL THEN 1 ELSE 0 END) AS null_historical_ctr,
    SUM(CASE WHEN is_high_intent_conversion IS NULL THEN 1 ELSE 0 END) AS null_label,
    ROUND(AVG(CAST(is_high_intent_conversion AS FLOAT64)), 4) AS label_positive_rate
FROM search_intelligence_db.search_facts;
"""

print("Data Quality & Null Verification Check running...")
```

## 4. Data limits

*What can this data never tell you? Unbalanced history, GSC-only early rows, window overlaps.*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
# Check for data density changes or missing feature gaps across time
query_limitations_check = """
SELECT
    DATE_TRUNC(event_date, MONTH) AS month_bucket,
    COUNT(*) AS total_rows,
    COUNT(DISTINCT query_id) AS distinct_queries,
    AVG(CASE WHEN historical_ctr IS NULL THEN 1.0 ELSE 0.0 END) AS missing_feature_ratio
FROM search_intelligence_db.search_facts
GROUP BY 1
ORDER BY 1 ASC;
"""

print("Checking temporal consistency and feature completeness over time...")
```
