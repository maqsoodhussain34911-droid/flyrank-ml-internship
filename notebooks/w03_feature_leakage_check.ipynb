## 1. Build the feature vector

*Code that actually builds it — engineered features, categorical handling, fills.*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd
import numpy as np

def build_features(df):
    # Example feature engineering steps
    df = df.copy()

    # 1. Handle missing values safely
    df['numeric_feature'] = df['numeric_feature'].fillna(df['numeric_feature'].median())
    df['categorical_feature'] = df['categorical_feature'].fillna('MISSING')

    # 2. Categorical handling (One-Hot Encoding or mapping)
    df = pd.get_dummies(df, columns=['categorical_feature'], drop_first=True)

    # 3. Engineered interaction feature (ensuring no future leakage)
    df['feature_ratio'] = df['feature_a'] / (df['feature_b'] + 1e-5)

    return df
```

## 2. Feature notes (meaning, missing, categorical, available-when?)

*For each feature: what it means, how missing values are handled, and whether it exists BEFORE the moment you predict.*

- **feature_a (Numeric):** Represents historical user activity count. Missing values are imputed using the median of the training set. Available *before* prediction time (lagged by 24 hours).
- **categorical_feature (Categorical):** Represents user segment or device type. Missing values are grouped into a dedicated 'MISSING' category to preserve information. Available *before* prediction time.
