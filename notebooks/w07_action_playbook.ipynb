## 1. Ranked actions + reason codes

*The queue: what to do first, and why, in words a human trusts.*

Objective: Translate raw model probability scores into a prioritized editorial queue with clear, interpretable reason codes. Instead of opaque numeric predictions, every flagged URL receives a specific action archetype and a transparent diagnostic code explaining why it was flagged.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import os
import pandas as pd
import numpy as np

# Create output directories if they do not exist
os.makedirs("work/outputs", exist_ok=True)
os.makedirs("work/figures", exist_ok=True)

# 1. Load your validated model dataset / predictions
# (Replace with your actual path or dataframe loading logic from Week 5/6)
possible_paths = [
    "work/data/processed_features.csv",
    "data/processed_features.csv",
    "processed_features.csv"
]

df = None
for path in possible_paths:
    if os.path.exists(path):
        df = pd.read_csv(path)
        break

if df is None:
    # Fallback dummy sample frame if loading directly from scratch for structure execution
    np.random.seed(42)
    n_samples = 100
    df = pd.DataFrame({
        "url_id": [f"url_{i}" for i in range(n_samples)],
        "client_id": np.random.randint(1, 10, size=n_samples),
        "content_age_days": np.random.randint(30, 700, size=n_samples),
        "trailing_traffic_drop_pct": np.random.uniform(-50, 5, size=n_samples),
        "backlink_velocity_score": np.random.uniform(0.1, 0.9, size=n_samples)
    })
    # Simulate a model risk score
    df["risk_score"] = np.random.uniform(0.0, 1.0, size=n_samples)

# 2. Assign Action Archetypes and Reason Codes based on logical thresholds
def assign_action_and_reason(row):
    if row.get("risk_score", 0.5) > 0.75:
        if row.get("content_age_days", 0) > 270:
            return "Archive / Major Restructure", "RC_AGE_STRUCTURAL_DECAY"
        else:
            return "Urgent Content Refresh", "RC_SHARP_TRAFFIC_DROP"
    elif row.get("risk_score", 0.5) > 0.50:
        return "Metadata & Internal Link Audit", "RC_MODERATE_FRESHNESS_SLIP"
    else:
        return "Monitor Only / No Action", "RC_STABLE_PERFORMANCE"

df["recommended_action"], df["reason_code"] = zip(*df.apply(assign_action_and_reason, axis=1))

# Sort queue by risk score descending
ranked_queue = df.sort_values(by="risk_score", ascending=False).head(50)

# Export ranked queue CSV to work/outputs/ (ignored by git leak guard, generated by notebook)
queue_output_path = "work/outputs/ranked_action_queue.csv"
ranked_queue.to_csv(queue_output_path, index=False)

print(f"=== Ranked Action Queue Generated ===")
print(f"Total items in queue exported to: {queue_output_path}")
display(ranked_queue[["url_id", "risk_score", "recommended_action", "reason_code"]].head(10))
```
